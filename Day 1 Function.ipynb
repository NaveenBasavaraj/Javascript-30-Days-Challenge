## Day 1 Create Hello World Function

Write a function `createHelloWorld`. It should return a new function that always returns `"Hello World"`.

**Example 1:**

* **Input:** `args = []`
* **Output:** `"Hello World"`
* **Explanation:**
```javascript
const f = createHelloWorld();
f(); // "Hello World"

```


The function returned by `createHelloWorld` should always return `"Hello World"`.

**Example 2:**

* **Input:** `args = [{}, null, 42]`
* **Output:** `"Hello World"`
* **Explanation:**
```javascript
const f = createHelloWorld();
f({}, null, 42); // "Hello World"

```


Any arguments could be passed to the function, but it should still always return `"Hello World"`.

## Solution

````typescript
// solution
function helloWorldFunc() {
    return function() {
        return "Hello World";
    };
};

var createHelloWorld = helloWorldFunc;
console.log('createHelloWorld is of type: ', typeof createHelloWorld)
const f = createHelloWorld();
console.log('f is of type: ', typeof f)

console.log(f());
````

````typescript
// solution 2

var createHelloWorld = function () {
    return function (){
        return "Hello World";
    };
};

const f = createHelloWorld(); // returns inner function to the variable

console.log(f())
````

````typescript
// solution 2

var createHelloWorld = () => {
    return () => {
        return "Hello World!"
    };
};

const f = createHelloWorld();

console.log(f());
````

### Concept 1 - Functions are Objects

Functions are **First Class Citizens**

That means they can:
* be stored in variables
* be passed to another function
* be returned from another function

````typescript
function greet(name) {
    return `Hello ${name}`;
}

const x = greet;

console.log(x('Naveen'));
````

We copied the whole function to a variable. Not the result.

### Concept 2 - Function expression


1. Function Declaration
- must function keyword
- must have explicit function name
- can be used be calling functionname()
- can also be assigned to a variable
- can be reused any number of times

````typescript
function addToNum(a,b) {
    return a + b;
};

const sumOf = addToNum(2,3)
sumOf
````

2. Function Expression
- function on the RHS does not have a name
- assigned a variable
- can be called using the variable

````typescript
const addTwoNum = function(a,b) {
    return a+b;
}

console.log(addTwoNum(2,5))
console.log(addTwoNum(2,50))
````

### Concept 3 — Returning Functions

Functions can return another function.

Example

```javascript
function outer() {

    return function () {
        return 5;
    };

}
```

Usage

```javascript
const x = outer();

console.log(x());
```

Output

```
5
```

### Concept 4 — Why return a function?

Sometimes you don't want to compute something immediately.

You want to prepare a function for later.

Example:

```javascript
function multiplyBy(x) {

    return function(y){
        return x * y;
    }

}

const double = multiplyBy(2);

console.log(double(8));
```

Output

```
16
```

## Practical Rule

* **No captured (free) variables** → just a function.
* **Captures variables from an outer scope** → closure.

Example:

```javascript
function outer() {
    let x = 10;

    return function () {
        return x;
    };
}
```

The inner function uses `x`, which isn't defined inside it. It is a **free variable**, and the function **closes over** `x`. This is what people usually mean by a closure.

---

## Technical JavaScript Definition

Technically, every function in JavaScript is created with a lexical environment (its closure), even if it doesn't actually use any outer variables.

For example:

```javascript
function outer() {
    return function () {
        return 42;
    };
}
```

The returned function still has an associated lexical environment, but it doesn't capture any variables from it. So, in everyday conversation, developers simply call it **a function**, not "a closure."
