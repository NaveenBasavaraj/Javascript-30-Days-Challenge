## Day 5 Apply Transform Over Each Element in Array



---

## 🎯 problem

Implement:

```javascript
var map = function(arr, fn) {
    
};
```

It receives:

* `arr` — an array of numbers
* `fn` — a function that transforms each element

It should return a **new array** containing the transformed values.

### Example

```javascript
arr = [1, 2, 3]
fn = function(x) {
    return x * 2;
}
```

Expected:

```javascript
[2, 4, 6]
```

Another:

```javascript
arr = [1, 2, 3]
fn = function(x, i) {
    return x + i;
}
```

Expected:

```javascript
[1, 3, 5]
```

Notice that `fn` can receive **both the value and its index**.

---

### Constraints

For this exercise, **don't use JavaScript's built-in `Array.prototype.map()`**.

Build the behavior yourself.

### Your task

Write your own:

```javascript
var map = function(arr, fn) {
    // your implementation
};
```



#### Solution

````typescript
var map = function(arr, fn) {
    const result = [];
    for (let i=0; i<arr.length; i++) {
        result.push(fn(arr[i], i));
    };
    return result;
};

console.log(map([1,2,3], (x,i) => x+i))
````
